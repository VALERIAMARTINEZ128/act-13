```python
class CAT:
    def __init__(self):
        # Los 5 Atributos
        self.Monto_Prestamo = 0.0
        self.Tasa_Interes = 0.0
        self.Plazo_Meses = 0
        self.Porcentaje_CAT = 0.0
        self.Nivel_Riesgo = "No Calculado"

    # --- LAS 5 FUNCIONES ---

    def configurar_Credito(self):
        print("\n--- 📝 [Función 1] CONFIGURAR CRÉDITO COMPLETO ---")
        self.Monto_Prestamo = float(input("¿Cuánto dinero necesitas? ($): "))
        self.Tasa_Interes = float(input("Ingresa la Tasa de Interés Anual (%): "))
        self.Plazo_Meses = int(input("¿A cuántos meses de plazo?: "))
        print("✅ Datos base configurados.")

    def calcular_CAT_Estimado(self):
        print("\n--- ⚙️ [Función 2] CALCULAR CAT ESTIMADO ---")
        if self.Tasa_Interes == 0:
            print("⚠️ Alerta: La tasa actual es 0%. Define una tasa primero.")
        # Simulación del CAT agregando una comisión fija de apertura y seguros (5.5%)
        self.Porcentaje_CAT = self.Tasa_Interes + 5.5
        print(f"📊 CAT Estimado Calculado: {self.Porcentaje_CAT}%")

    def analizar_Riesgo_IA(self):
        print("\n--- 🤖 [Función 3] ANALIZAR RIESGO CON IA ---")
        if self.Porcentaje_CAT == 0:
            print("❌ No se puede analizar. El CAT actual es 0%. Ejecuta la función 2 primero.")
            return

        if self.Porcentaje_CAT < 20:
            self.Nivel_Riesgo = "Bajo (¡Excelente oportunidad!)"
        elif self.Porcentaje_CAT >= 20 and self.Porcentaje_CAT <= 50:
            self.Nivel_Riesgo = "Medio (Modera tus gastos)"
        else:
            self.Nivel_Riesgo = "Alto (⚠️ ¡Peligro! Es un crédito muy costoso)"
        print(f"🧠 IA: Análisis de riesgo establecido como: {self.Nivel_Riesgo}")

    def simular_Mensualidad(self):
        print("\n--- 💸 [Función 4] SIMULAR MENSUALIDAD ---")
        if self.Monto_Prestamo == 0 or self.Plazo_Meses == 0:
            print("❌ Datos insuficientes. Asegúrate de tener Monto y Plazo mayores a cero.")
            return
        interes_estimado = self.Monto_Prestamo * (self.Tasa_Interes / 100)
        pago_total = self.Monto_Prestamo + interes_estimado
        cuota = pago_total / self.Plazo_Meses
        print(f"💰 Tu pago estimado sería de: ${cuota:,.2f} al mes por {self.Plazo_Meses} meses.")

    def mostrar_Resumen(self):
        print("\n=============================================")
        print("    📊 [Función 5] RESUMEN GENERAL DEL CAT   ")
        print("=============================================")
        print(f" 1. Monto del Préstamo:   ${self.Monto_Prestamo:,.2f}")
        print(f" 2. Tasa de Interés:      {self.Tasa_Interes}%")
        print(f" 3. Plazo en Meses:       {self.Plazo_Meses} meses")
        print(f" 4. Porcentaje de CAT:    {self.Porcentaje_CAT}%")
        print(f" 5. Nivel de Riesgo IA:   {self.Nivel_Riesgo}")
        print("=============================================")


# --- MENÚ INTERACTIVO DE 10 OPCIONES ---
def ejecutor():
    ia_cat = CAT()

    while True:
        print("\n==================================================")
        print("       🤖 SISTEMA DE CONTROL DE CAT (10 OPCIONES)   ")
        print("==================================================")
        print(" [ SECCIÓN ATRIBUTOS: Ver / Modificar Datos ]")
        print("  1. Modificar Atributo: Monto_Prestamo")
        print("  2. Modificar Atributo: Tasa_Interes")
        print("  3. Modificar Atributo: Plazo_Meses")
        print("  4. Consultar Atributo: Porcentaje_CAT")
        print("  5. Consultar Atributo: Nivel_Riesgo")
        print(" ------------------------------------------------")
        print(" [ SECCIÓN FUNCIONES: Ejecutar Procesos de la IA ]")
        print("  6. Ejecutar Función: configurar_Credito()")
        print("  7. Ejecutar Función: calcular_CAT_Estimado()")
        print("  8. Ejecutar Función: analizar_Riesgo_IA()")
        print("  9. Ejecutar Función: simular_Mensualidad()")
        print("  10. Ejecutar Función: mostrar_Resumen()")
        print(" ------------------------------------------------")
        print("  0. Salir del Sistema")

        opcion = input("\nSelecciona una opción (0-10): ")

        # --- Lógica de los Atributos ---
        if opcion == "1":
            ia_cat.Monto_Prestamo = float(input("\n[Atributo 1] Ingresa nuevo Monto del Préstamo $: "))
            print("✅ Atributo actualizado.")
        elif opcion == "2":
            ia_cat.Tasa_Interes = float(input("\n[Atributo 2] Ingresa nueva Tasa de Interés %: "))
            print("✅ Atributo actualizado.")
        elif opcion == "3":
            ia_cat.Plazo_Meses = int(input("\n[Atributo 3] Ingresa nuevo Plazo en meses: "))
            print("✅ Atributo actualizado.")
        elif opcion == "4":
            print(f"\n🔍 [Atributo 4] Valor actual de Porcentaje_CAT: {ia_cat.Porcentaje_CAT}%")
        elif opcion == "5":
            print(f"\n🔍 [Atributo 5] Valor actual de Nivel_Riesgo: {ia_cat.Nivel_Riesgo}")

        # --- Lógica de las Funciones ---
        elif opcion == "6":
            ia_cat.configurar_Credito()
        elif opcion == "7":
            ia_cat.calcular_CAT_Estimado()
        elif opcion == "8":
            ia_cat.analizar_Riesgo_IA()
        elif opcion == "9":
            ia_cat.simular_Mensualidad()
        elif opcion == "10":
            ia_cat.mostrar_Resumen()

        # --- Salida ---
        elif opcion == "0":
            print("\n👋 Sistema cerrado. ¡Hasta luego!")
            break
        else:
            print("\n❌ Opción incorrecta. Introduce un número del 0 al 10.")

# Iniciar el programa simulador
ejecutor()
```
